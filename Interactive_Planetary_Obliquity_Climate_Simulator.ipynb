```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# Define physics parameters for 3 Earth regions (Latitude in degrees)
LATITUDE_EQUATOR = 0
LATITUDE_NORTH_POLE = 90
LATITUDE_SOUTH_POLE = -90

def simulate_climate_response(axial_tilt=23.5):
    """
    Simulates planetary tilt changes and recalculates temperatures
    for Equator, North Pole, and South Pole based on Lambert's Cosine Law.
    """
    # Create a layout with two subplots (Left: Planet Visualization | Right: Temperature Response)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    tilt_rad = np.radians(axial_tilt)

    # ----------------------------------------------------
    # LEFT PLOT: Interactive 2D Structure of Earth & Rotational Axis
    # ----------------------------------------------------
    # Orbital ecliptic plane (Sun rays reference line)
    ax1.plot([-1.8, 1.8], [0, 0], color='gold', lw=1.5, linestyle='--', label='Orbital Plane')

    # Draw Earth as a blue circle
    earth = plt.Circle((0, 0), 1.0, color='royalblue', alpha=0.8, edgecolor='k', lw=1)
    ax1.add_patch(earth)

    # Draw the dynamically tilted Rotational Axis (North and South)
    ax1.plot([-1.4 * np.sin(tilt_rad), 1.4 * np.sin(tilt_rad)],
             [-1.4 * np.cos(tilt_rad), 1.4 * np.cos(tilt_rad)],
             color='crimson', lw=3.5, label=f'Rotational Axis ({axial_tilt:.1f}°)')

    # Label the Poles directly on the moving axis vectors
    ax1.text(1.5 * np.sin(tilt_rad), 1.5 * np.cos(tilt_rad), 'N', fontsize=12, weight='bold', color='crimson', ha='center')
    ax1.text(-1.5 * np.sin(tilt_rad), -1.5 * np.cos(tilt_rad), 'S', fontsize=12, weight='bold', color='crimson', ha='center')

    # Draw incoming Solar Radiation vectors (Rays from the left)
    for y_pos in [-0.6, 0, 0.6]:
        ax1.quiver(-1.9, y_pos, 0.5, 0, scale=3, color='orange', alpha=0.8, width=0.008)

    ax1.set_xlim(-2.2, 2.2)
    ax1.set_ylim(-2.2, 2.2)
    ax1.set_title('Earth Axial Obliquity Visualization', fontsize=12, weight='bold')
    ax1.set_aspect('equal')
    ax1.axis('off')
    ax1.legend(loc='upper right')

    # ----------------------------------------------------
    # RIGHT PLOT: Temperature Metrics across 3 Regions
    # ----------------------------------------------------
    # Physics formula modeling thermal energy deltas based on latitude exposure to the tilt
    tilt_range = np.linspace(0, 45, 100)

    # Simulating long-term trend curves for the chart background
    temp_equator_curve = 32 - 0.15 * (tilt_range - 23.5)**2
    temp_north_curve = -25 + 1.4 * (tilt_range - 23.5)
    temp_south_curve = -28 + 1.2 * (tilt_range - 23.5)

    # Current active temperature values based on user input selection
    current_eq_temp = 32 - 0.15 * (axial_tilt - 23.5)**2
    current_np_temp = -25 + 1.4 * (axial_tilt - 23.5)
    current_sp_temp = -28 + 1.2 * (axial_tilt - 23.5)

    # Chart baseline trendlines
    ax2.plot(tilt_range, temp_equator_curve, color='darkred', alpha=0.3, linestyle=':')
    ax2.plot(tilt_range, temp_north_curve, color='teal', alpha=0.3, linestyle=':')
    ax2.plot(tilt_range, temp_south_curve, color='navy', alpha=0.3, linestyle=':')

    # Highlight current interactive values with large markers
    ax2.scatter(axial_tilt, current_eq_temp, color='darkred', s=120, edgecolors='k', zorder=5, label=f'Equator: {current_eq_temp:.1f}°C')
    ax2.scatter(axial_tilt, current_np_temp, color='teal', s=120, edgecolors='k', zorder=5, label=f'North Pole: {current_np_temp:.1f}°C')
    ax2.scatter(axial_tilt, current_sp_temp, color='navy', s=120, edgecolors='k', zorder=5, label=f'South Pole: {current_sp_temp:.1f}°C')

    ax2.set_xlim(0, 45)
    ax2.set_ylim(-60, 45)
    ax2.set_xlabel('Axial Tilt Angle (Degrees)', fontsize=10)
    ax2.set_ylabel('Mean Regional Temperature (°C)', fontsize=10)
    ax2.set_title('Synchronous Regional Climate Feedback', fontsize=12, weight='bold')
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.legend(loc='lower left', fontsize=10)

    plt.tight_layout()
    plt.show()

# 3. Create interactive Sliders for the UI controller (Allow tilt values from 0 to 45 degrees)
interact(simulate_climate_response,
         axial_tilt=widgets.FloatSlider(value=23.5, min=0.0, max=45.0, step=0.5,
                                       description="Axial Tilt (°):", continuous_update=True))
```
